# Line Magics (`%`) — Una línea

En Jupyter, los **comandos mágicos** son atajos que empiezan con `%` (una celda) o `%%` (toda la celda).

Este notebook explica los **line magics** personalizados del taller: un comando por línea, sin escribir funciones largas.

### Cómo usarlos

1. Ejecuta primero la celda de **carga** (abajo).
2. En cualquier celda de código, escribe `%comando` o `%comando argumento`.
3. Ejecuta la celda con **Shift+Enter**.

| Comando | Uso | Ejemplo |
|---------|-----|--------|
| `%memory_usage` | Ver RAM usada | `%memory_usage` |
| `%env_vars` | Variables de entorno | `%env_vars` |
| `%show_pwd` | Directorio actual | `%show_pwd` |
| `%check_aws_creds` | Verificar AWS | `%check_aws_creds` |
| `%check_gcp_creds` | Verificar GCP | `%check_gcp_creds` |
| `%check_azure_creds` | Verificar Azure | `%check_azure_creds` |
| `%df_info` | Info de DataFrame | `%df_info df` |
| `%df_preview` | Preview de DataFrame | `%df_preview df` |
| `%clear_cache` | Limpiar caché de módulos | `%clear_cache` |
| `%timestamp` | Hora actual | `%timestamp` |
| `%trace_var` | Inspeccionar variable | `%trace_var mi_var` |
| `%list_vars` | Listar variables | `%list_vars` |

## Cargar los magics

Ejecuta esta celda **una vez** al abrir el notebook. Registra todos los comandos `%` en el kernel.

```python
import sys
from pathlib import Path

from IPython import get_ipython

sys.path.insert(0, str(Path("src").resolve()))
import line_magics

line_magics.load_ipython_extension(get_ipython())

print("Line magics cargados. Prueba: %timestamp")
```

---

## 1. `%memory_usage` — RAM del proceso

Muestra cuánta memoria ha usado **este kernel** de Jupyter. Útil para detectar notebooks que consumen mucha RAM.

```python
%memory_usage
```

---

## 2. `%env_vars` — Variables de entorno

Sin argumentos, lista todas las variables de entorno. Con un nombre, muestra solo esa variable.

```python
# Todas las variables (puede ser largo)
# %env_vars

# Una variable concreta
%env_vars PATH
```

---

## 3. `%show_pwd` — Directorio actual

Equivale a `pwd` en la terminal: dónde está ejecutándose el notebook.

```python
%show_pwd
```

---

## 4. Credenciales en la nube

Estos tres magics **no conectan** a la nube: solo comprueban si hay señales de credenciales configuradas (variables de entorno o archivos habituales).

```python
%check_aws_creds
```

```python
%check_gcp_creds
```

```python
%check_azure_creds
```

---

## 5. `%df_info` y `%df_preview` — DataFrames

Trabajan con un `pandas.DataFrame` que ya hayas creado. El argumento es el **nombre de la variable**.

```python
import pandas as pd

df = pd.DataFrame({
    "acorde": ["C", "G", "Am", "F"],
    "notas": ["C-E-G", "G-B-D", "A-C-E", "F-A-C"],
    "trastes": [0, 3, 0, 1],
})

df
```

```python
%df_info df
```

```python
%df_preview df
```

---

## 6. `%clear_cache` — Recargar módulos

Si editas un archivo `.py` en `src/` mientras el notebook sigue abierto, Python sigue usando la versión antigua en memoria. Este magic **recarga** los módulos del proyecto.

```python
%clear_cache
```

---

## 7. `%timestamp` — Hora actual

Marca el momento en que ejecutas una celda (útil en logs o experimentos).

```python
%timestamp
```

---

## 8. `%trace_var` — Inspeccionar una variable

Muestra tipo, valor y tamaño (si aplica) de cualquier variable del notebook.

```python
mi_var = ["E", "A", "D", "G", "B", "E"]

%trace_var mi_var
```

---

## 9. `%list_vars` — Variables definidas

Lista las variables que **tú** has creado en el notebook (ignora variables internas de IPython).

```python
%list_vars
```

---

## Resumen

| Magic | ¿Necesita argumento? | Cuándo usarlo |
|-------|----------------------|---------------|
| `%memory_usage` | No | Diagnóstico de memoria |
| `%env_vars` | Opcional | Revisar configuración del entorno |
| `%show_pwd` | No | Saber desde qué carpeta corre el kernel |
| `%check_*_creds` | No | Comprobar si hay credenciales cloud |
| `%df_info` / `%df_preview` | Sí (nombre del DataFrame) | Explorar datos tabulares |
| `%clear_cache` | No | Tras editar código en `src/` |
| `%timestamp` | No | Registrar cuándo corrió algo |
| `%trace_var` | Sí (nombre de variable) | Depurar un valor concreto |
| `%list_vars` | No | Ver qué hay en memoria |

> **Tip:** Los magics son atajos. Bajo el capó llaman a funciones Python en [`src/line_magics.py`](src/line_magics.py).
