## 1. Funciones en 30 segundos

Una **función** es un bloque reutilizable con un nombre. Recibe datos (**parámetros**) y puede devolver un resultado (`return`).

En el taller ya usaste funciones como `buscar_nota()` y `dibujar_mastil()`. Aquí repasamos la idea con un ejemplo mínimo.

```python
def nombre_acorde(notas: list[str]) -> str:
    """Une las notas con guiones para mostrarlas."""
    return " - ".join(notas)


do_mayor = ["C", "E", "G"]
print(nombre_acorde(do_mayor))
```

**Ventajas:** evitas repetir código, das nombre a la intención y puedes probar cada pieza por separado.

## 2. SRP — una función, una responsabilidad

**SRP** (*Single Responsibility Principle*): cada función debería hacer **una sola cosa** y hacerla bien.

En `workshop_hands_on.ipynb` separamos dos tareas distintas:

| Función | Responsabilidad | Qué **no** debería hacer |
|---------|-----------------|--------------------------|
| `obtener_color()` | Decidir el color de una posición (cuerda, traste) | Dibujar el mástil |
| `dibujar_mastil()` | Generar la figura con matplotlib | Saber la lógica de colores del acorde |

Si mezcláramos todo en una sola función gigante, cambiar el color de un acorde obligaría a tocar también el dibujo del mástil.

```python
ACORDE_C = {
    "nombre": "C",
    "posiciones": [
        {"cuerda": 1, "traste": 0},
        {"cuerda": 2, "traste": 1},
        {"cuerda": 4, "traste": 2},
    ],
    "color_resaltado": "#E53935",
}


def obtener_color(cuerda: int, traste: int, acorde: dict | None, color_default: str = "black") -> str:
    if acorde is None:
        return color_default
    for pos in acorde["posiciones"]:
        if pos["cuerda"] == cuerda and pos["traste"] == traste:
            return acorde.get("color_resaltado", "red")
    return color_default


# Solo decide colores — no dibuja nada
print(obtener_color(2, 1, ACORDE_C))   # posición del acorde → rojo
print(obtener_color(0, 5, ACORDE_C))   # fuera del acorde → negro
```

`dibujar_mastil()` **delega** el color: llama a `obtener_color()` por cada casilla. Así cada función tiene un motivo claro para cambiar.

## 3. OCP — extender sin romper

**OCP** (*Open/Closed Principle*): el código debería estar **abierto a extensión** y **cerrado a modificación**.

Ejemplo: queremos resaltar también las notas al aire (traste 0) con otro color. En lugar de llenar `dibujar_mastil()` de `if`, podemos **cambiar solo** `obtener_color()`:

```python
def obtener_color(cuerda: int, traste: int, acorde: dict | None, color_default: str = "black") -> str:
    if acorde is None:
        return color_default
    if traste == 0:
        return "#90CAF9"  # azul claro para cuerdas al aire
    for pos in acorde["posiciones"]:
        if pos["cuerda"] == cuerda and pos["traste"] == traste:
            return acorde.get("color_resaltado", "red")
    return color_default


print("Al aire:", obtener_color(1, 0, ACORDE_C))
print("En acorde:", obtener_color(2, 1, ACORDE_C))
```

`dibujar_mastil()` sigue igual: solo necesita *alguna* función que devuelva un color. Eso es extensión sin reescribir el dibujador.

## 4. El resto de SOLID (solo una pincelada)

SOLID son cinco principios de diseño orientado a objetos. En este taller solo aplicamos bien **S** y **O**. Los otros tres los verás con calma más adelante:

| Letra | Nombre | Idea en una frase |
|-------|--------|-------------------|
| **L** | Liskov Substitution | Los subtipos deben poder sustituir al tipo base sin sorpresas |
| **I** | Interface Segregation | Interfaces pequeñas, no contratos enormes |
| **D** | Dependency Inversion | Depender de abstracciones, no de detalles concretos |

No hace falta memorizarlos ahora. Lo importante del taller: **funciones pequeñas, nombres claros, una responsabilidad cada una**.
