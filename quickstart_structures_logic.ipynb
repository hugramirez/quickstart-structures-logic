## Listas

Un **acorde** son varias notas musicales. En Python las guardamos en una **lista**: elementos en orden, separados por comas, entre corchetes `[ ]`.

> Analogía: una lista es una fila numerada. El puesto **0** es el primero (no el 1).

### Crear una lista de notas

```python
notas_do_mayor = ["C", "E", "G"]

print(notas_do_mayor)
print("Tipo:", type(notas_do_mayor))
```

### Índices: acceder a cada nota

Los índices empiezan en **0**:

| Índice | Nota |
|--------|------|
| `[0]` | C (Do) |
| `[1]` | E (Mi) |
| `[2]` | G (Sol) |

```python
print("Primera nota [0]:", notas_do_mayor[0])
print("Segunda nota [1]:", notas_do_mayor[1])
print("Tercera nota [2]:", notas_do_mayor[2])
print("Última nota [-1]:", notas_do_mayor[-1])
```

### `len()`: cuántas notas tiene el acorde

```python
print(f"Do Mayor tiene {len(notas_do_mayor)} notas.")
```

### Modificar y añadir notas

Podemos cambiar una posición o añadir una nota al final con `.append()`.

```python
notas_la_menor = ["A", "C", "E"]
print("Antes:", notas_la_menor)

notas_la_menor[1] = "C#"  # variante con C sostenido
print("Tras cambiar [1]:", notas_la_menor)

notas_la_menor[1] = "C"  # restauramos La menor
notas_la_menor.append("A")  # octava grave
print("Tras append:", notas_la_menor)
```

### Recorrer la lista con `for`

Imprimimos cada nota del acorde una a una.

```python
print("Notas de Do Mayor:")
for nota in notas_do_mayor:
    print(f"  - {nota}")
```

### ¿Está una nota en el acorde?

El operador `in` devuelve `True` o `False`.

```python
print("¿Mi está en Do Mayor?", "E" in notas_do_mayor)
print("¿Fa está en Do Mayor?", "F" in notas_do_mayor)
```

### Varios acordes: lista de listas

Cada acorde es una lista; todos juntos forman otra lista.

```python
acordes_simples = [
    ["C", "E", "G"],   # Do Mayor
    ["A", "C", "E"],   # La menor
    ["G", "B", "D"],   # Sol Mayor
]

print("Segundo acorde:", acordes_simples[1])
print("Primera nota del segundo acorde:", acordes_simples[1][0])
```

### Errores comunes — Listas

- **`IndexError`**: pedir `[3]` en una tríada (solo existen índices 0, 1 y 2).
- **Confundir posición humana con índice**: la 1.ª nota es `[0]`, no `[1]`.

## Diccionarios

Una lista guarda notas en orden, pero **no nombra** cada dato. Un **diccionario** usa pares **clave → valor**, como una ficha de acorde:

- `"nombre"` → `"C"`
- `"notas"` → `["C", "E", "G"]`

> Analogía: el diccionario es la ficha del acorde; la lista de notas vive *dentro* de la ficha.

### Crear un acorde como diccionario

```python
acorde_do_mayor = {
    "nombre": "C",
    "nombre_completo": "Do Mayor",
    "notas": ["C", "E", "G"],
    "color": "#E53935",
}

print(acorde_do_mayor)
```

### Leer campos del diccionario

Accedemos con la clave entre corchetes: `acorde["nombre"]`.

```python
print("Nombre corto:", acorde_do_mayor["nombre"])
print("Nombre completo:", acorde_do_mayor["nombre_completo"])
print("Notas:", acorde_do_mayor["notas"])
print("Primera nota:", acorde_do_mayor["notas"][0])  # lista dentro del dict
```

### Añadir y cambiar campos

```python
acorde_do_mayor["digitacion"] = "x-3-2-0-1-0"
acorde_do_mayor["color"] = "#C62828"

print("Digitación:", acorde_do_mayor["digitacion"])
print("Color:", acorde_do_mayor["color"])
```

### Recorrer un diccionario

Con `.items()` obtenemos cada par clave → valor.

```python
print(f"Ficha de {acorde_do_mayor['nombre_completo']}:")
for clave, valor in acorde_do_mayor.items():
    print(f"  {clave}: {valor}")
```

### ¿Existe una clave?

Usa `in` sobre las claves antes de acceder, o `.get()` para evitar errores.

```python
print('"notas" in acorde?', "notas" in acorde_do_mayor)
print('"bpm" in acorde?', "bpm" in acorde_do_mayor)
print("BPM (con get):", acorde_do_mayor.get("bpm", "no definido"))
```

### Varios acordes: diccionario de diccionarios

La **clave** (`"C"`, `"Am"`…) identifica cada acorde. En el notebook del mástil usaremos la misma idea con más campos (`posiciones`, `color_resaltado`).

```python
acordes = {
    "C": {
        "nombre": "C",
        "nombre_completo": "Do Mayor",
        "notas": ["C", "E", "G"],
        "color": "#E53935",
    },
    "Am": {
        "nombre": "Am",
        "nombre_completo": "La menor",
        "notas": ["A", "C", "E"],
        "color": "#1E88E5",
    },
    "G": {
        "nombre": "G",
        "nombre_completo": "Sol Mayor",
        "notas": ["G", "B", "D"],
        "color": "#43A047",
    },
}

print("La menor:", acordes["Am"]["nombre_completo"])
print("Notas:", acordes["Am"]["notas"])
```

### Recorrer todos los acordes

```python
for clave, acorde in acordes.items():
    notas = " - ".join(acorde["notas"])
    print(f"{clave}: {acorde['nombre_completo']} → {notas}")
```

### Errores comunes — Diccionarios

- **`KeyError`**: clave mal escrita (`"nota"` en lugar de `"notas"`).
- **Confundir clave y valor**: `"C"` es la clave del diccionario grande; `"C"` dentro de `"notas"` es una nota musical.

## Resumen — Lista vs Diccionario

| Estructura | Piensa en… | Sintaxis clave | Ejemplo con acordes |
|------------|------------|----------------|---------------------|
| **Lista** | Notas en orden | `[ ]`, `[0]`, `for n in lista` | `["C", "E", "G"]` |
| **Diccionario** | Ficha del acorde | `{ }`, `["clave"]`, `.items()` | `{"nombre": "C", "notas": [...]}` |

**Combinación habitual:** `acorde["notas"][0]` — diccionario que contiene una lista.
